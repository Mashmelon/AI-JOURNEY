```python
# ================================
# AIR CARGO AI CHATBOT (COLAB VERSION)
# ================================

import pandas as pd

# -------------------------------
# 1. SAMPLE DATA (You can replace later with CSV)
# -------------------------------
data = {
    "question": [
        "What is air cargo?",
        "How to track shipment?",
        "What is airway bill?",
        "Delivery time for cargo?",
        "Contact support",
        "How to cancel shipment?"
    ],
    "answer": [
        "Air cargo refers to goods transported by aircraft.",
        "You can track shipment using tracking ID on our website.",
        "Airway bill is a document issued by airline for cargo.",
        "Delivery usually takes 3-5 business days.",
        "You can contact support at support@aircargo.com",
        "Shipment can be cancelled before dispatch."
    ],
    "category": [
        "general",
        "tracking",
        "general",
        "delivery",
        "support",
        "cancel"
    ]
}

df = pd.DataFrame(data)

# -------------------------------
# 2. INTENT DETECTION
# -------------------------------
def detect_intent(query):
    query = query.lower()

    if "track" in query:
        return "tracking"
    elif "delivery" in query or "time" in query:
        return "delivery"
    elif "cancel" in query:
        return "cancel"
    elif "support" in query or "contact" in query:
        return "support"
    else:
        return "general"

# -------------------------------
# 3. SEARCH FUNCTION (FIXED INDENTATION)
# -------------------------------
def search_answer(query, df_filtered):
    query_words = query.lower().split()

    for q, a in zip(df_filtered['question'], df_filtered['answer']):
        if any(word in q.lower() for word in query_words):
            return a

    return None

# -------------------------------
# 4. CHATBOT FUNCTION
# -------------------------------
def chatbot(query):
    intent = detect_intent(query)

    if intent != "general":
        df_filtered = df[df['category'] == intent]
    else:
        df_filtered = df

    answer = search_answer(query, df_filtered)

    if not answer:
        return "We could not find an exact answer. Please contact support."

    return f"""
Dear Customer,

{answer}

Regards,
Air Cargo AI Assistant
"""

# -------------------------------
# 5. TEST CHATBOT
# -------------------------------
print("✈️ Air Cargo AI Chatbot (type 'exit' to stop)\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Thank you! Goodbye 👋")
        break

    response = chatbot(user_input)
    print("Bot:", response)
```
